## 1. Ryser's Exchange for Class 2 Generation
We use a checkerboard swap (Ryser's exchange) on rows of the *same sum* to generate non-isomorphic incidence matrices that preserve exact row-sums and column-sums.

```
# Generate a random base incidence matrix B_x
np.random.seed(42)
random.seed(42)

k = 6  # 6 distinct shared neighbors
m = 8  # 8 incident hyperedges

B_x = generate_base_matrix(k, m)
print("Original B_x (Class 3 Base):")
print(B_x)

# Apply Ryser's Exchange to get B_y for Class 2
B_y_class2 = apply_ryser_exchange(B_x)
print("\nModified B_y (Class 2 Partner):")
print(B_y_class2)
```

## 2. Verify Mathematical Invariances
Let's prove that this transformation preserves degrees and hyperedge sizes, but destroys strong structural equivalence.

```
print("Row Sums (Hyperedge sizes) Match:", np.array_equal(sorted(B_x.sum(axis=1)), sorted(B_y_class2.sum(axis=1))))
print("Column Sums (Neighbor degrees) Match:", np.array_equal(B_x.sum(axis=0), B_y_class2.sum(axis=0)))

# Are they strongly equivalent (identical multiset of row supports)?
is_strong_equiv = strong_structural_equivalence_multiset(B_x, B_y_class2)
print("Strongly Equiv (Class 3)?", is_strong_equiv)
```
